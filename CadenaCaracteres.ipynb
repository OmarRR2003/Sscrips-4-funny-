Cadena de Caracteres

```python
# lIBRERIAS
from librerias import *
```

```python
# Definimos listas y variables

## Definimos el abecedario incluyendo el caracter de 'espacio' para poder separar palabras
abc = ['a','b','c','d','e','f','g','h','i','j','k','l','m','n','ñ','o','p','q','r','s','t','u','v','w','x','y','z', ' ']

## Definimos la letra de canción a analizar
cancion = ["""Que dejen toditos los libros abiertos,
ha sido la orden que dio el general.
Que todos los niños estén muy atentos,
las cinco vocales van a desfilar.

Primero verás que pasa la A,
con sus dos patitas muy abiertas al marchar;
ahí viene la E, alzando los pies,
el palo de enmedio es más chico, como ves;
aquí está la I, le sigue la O,
una flaca y otra gorda porque ya comió
y luego hasta atrás, llegó la U,
como la cuerda con que siempre saltas tú."""]
```

```python
# Estandarización de la letra

## Eliminamos caracteres que no pertenecen al abecedario y convertimos a minúsculas
cancion = cancion[0].lower().replace('\n', ' ').replace(',',' ').replace('.','').replace(';','')

## Quitamos acentos
def quitar_acentos(texto):
    # 1. Descompone los caracteres acentuados en letra base + acento (NFD)
    texto_normalizado = unicodedata.normalize('NFD', texto)
    
    # 2. Filtra y conserva solo los caracteres que NO sean acentos (categoría 'Mn')
    texto_sin_acentos = ''.join(c for c in texto_normalizado if unicodedata.category(c) != 'Mn')
    
    return texto_sin_acentos

texto_sin_acentos = quitar_acentos(cancion)

print("La canción estandarizada es:\n", texto_sin_acentos)
```

Funcion de caracteres

```python
def crear_matriz_transicion(texto, abc):
    # Crear un diccionario con conteos en 0 para todas las combinaciones posibles
    conteo = {c1: {c2: 0 for c2 in abc} for c1 in abc}
    
    #  Recorrer el texto contando el carácter actual y el siguiente
    for i in range(len(texto) - 1):
        actual = texto[i]
        siguiente = texto[i + 1]
        
        # Solo contamos si ambos caracteres pertenecen al abecedario definido
        if actual in conteo and siguiente in conteo[actual]:
            conteo[actual][siguiente] += 1
            
    #  Convertir el diccionario a un Df
    # T traspone para que las FILAS sean el carácter actual y las COLUMNAS el siguiente
    matriz_df = pd.DataFrame(conteo).T
    
    return matriz_df

# Generamos la matriz 
matriz_28x28 = crear_matriz_transicion(texto_sin_acentos, abc)


print("Dimensión de la matriz:", matriz_28x28.shape)
```
